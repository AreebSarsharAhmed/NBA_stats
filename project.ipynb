```python
# Importing necessary libraries
import pandas as pd
import matplotlib.pyplot as plt

# Section 1: Dataset Exploration
# 1. Load the dataset and display the first 10 rows
nba_df = pd.read_csv('C:\\Users\\91950\\Downloads\\nba_stats.csv')
print("First 10 rows of the dataset:")
print(nba_df.head(10))
```

```python
# 2. Display the shape of the dataset
print("\nShape of the dataset (rows, columns):", nba_df.shape)
```

```python
# 3. Identify data types and check for missing values
print("\nData types and missing values:")
print(nba_df.info())
```

```python
# 4. List the number of players who played for each team
print("\nNumber of players who played for each team:")
team_counts = nba_df['Tm'].value_counts()
print(team_counts)
```

```python
# Section 2: Data Cleaning and Preparation
# 1. Remove duplicate rows
nba_df = nba_df.drop_duplicates()
nba_df
```

```python
# 2. Rename columns 'Tm' to 'Team' and 'MP' to 'Mins_Played'
nba_df = nba_df.rename(columns={'Tm': 'Team', 'MP': 'Mins_Played'})
nba_df
```

```python
# 3. Remove rows with 'TOT' in the Team column
nba_df = nba_df[nba_df['Team'] != 'TOT']
nba_df
```

```python
# 4. Create a new column 'FT_PERCENT' for free throw percentage
nba_df['FT_PERCENT'] = (nba_df['FT'] / nba_df['FTA']) * 100
nba_df
```

```python
# 5. Create a new column 'TRB' for total rebounds
nba_df['TRB'] = nba_df['ORB'] + nba_df['DRB']
nba_df
```

```python
# 6. Convert 3P% and 2P% from decimal to percentage format
nba_df['3P%'] = nba_df['3P%'] * 100
nba_df['2P%'] = nba_df['2P%'] * 100
nba_df
```

```python
# Section 3: Data Analysis and Visualization
# 1. Count players who scored more than 2,000 points in the year 92-93
points_92_93 = nba_df[(nba_df['YEAR'] == '92-93') & (nba_df['PTS'] > 2000)]
print("\nNumber of players who scored more than 2,000 points in the year 92-93:", len(points_92_93))
```

```python
# 2. Create a line plot showing the frequency of the top 5 players with the most 3 Points made in the year 94-95
top_5_3P_94_95 = nba_df[nba_df['YEAR'] == '94-95'].nlargest(5, '3P')
plt.plot(top_5_3P_94_95['Player'], top_5_3P_94_95['3P'], marker='o')
plt.xlabel('Player')
plt.ylabel('3 Points Made')
plt.title('Top 5 Players with Most 3 Points Made in 94-95')
plt.xticks(rotation=45)
plt.show()
```

```python
# 3. What is the relationship between the player’s age and the number of points scored? 
# The scatter plot should show the player's age on the x-axis, and the number of points scored on the y-axis.
plt.figure(figsize=(10, 6))
plt.scatter(nba_df['Age'], nba_df['PTS'], alpha=0.5)
plt.title('Relationship between Age and Points Scored')
plt.xlabel('Age')
plt.ylabel('Points Scored')
plt.grid(True)
plt.show()
```

```python
# 4. Create a boxplot that shows the number of points scored for each position in the year 90-91.
plt.figure(figsize=(10, 6))
nba_90_91 = nba_df[nba_df['YEAR'] == '90-91']
plt.boxplot([nba_90_91[nba_90_91['Pos'] == pos]['PTS'] for pos in nba_90_91['Pos'].unique()],
            labels=nba_90_91['Pos'].unique())
plt.title('Number of Points Scored by Position in 90-91')
plt.xlabel('Position')
plt.ylabel('Points Scored')
plt.show()
```

```python
# 5. Create a bar chart that shows the total number of players by age in the year 90-91.
plt.figure(figsize=(10, 6))
players_by_age = nba_90_91['Age'].value_counts().sort_index()
plt.bar(players_by_age.index, players_by_age.values)
plt.title('Total Number of Players by Age in 90-91')
plt.xlabel('Age')
plt.ylabel('Number of Players')
plt.show()
```

```python
# 6. Create a bar chart that counts the number of players at each position in the year 90-91.
plt.figure(figsize=(10, 6))
players_by_position = nba_90_91['Pos'].value_counts()
plt.bar(players_by_position.index, players_by_position.values)
plt.title('Number of Players by Position in 90-91')
plt.xlabel('Position')
plt.ylabel('Number of Players')
plt.show()
```

```python
# 7. Who led the NBA in Total Rebounds in the year 91-92?
top_rebounder_91_92 = nba_df[nba_df['YEAR'] == '91-92'].nlargest(1, 'TRB')
print("\nPlayer who led the NBA in Total Rebounds in 91-92:", top_rebounder_91_92['Player'].values[0])
```

```python
# 8. Who had thehighest assist average in the year 93-94? What was the assist average for that player?
top_assist_avg_93_94 = nba_df[nba_df['YEAR'] == '93-94'].nlargest(1, 'AST')
print("\nPlayer with the highest assist average in 93-94:", top_assist_avg_93_94['Player'].values[0])
print("Assist average for that player:", top_assist_avg_93_94['AST'].values[0])
```
